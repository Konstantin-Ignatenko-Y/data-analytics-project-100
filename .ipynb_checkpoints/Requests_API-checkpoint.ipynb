## Запросы к API

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns
import requests
import plotly.graph_objects as go
import plotly.express as px
from dotenv import load_dotenv
```

```python
url_visits = "https://data-charts-api.hexlet.app/visits"
url_reg = "https://data-charts-api.hexlet.app/registrations"

params = {
    "begin": "2023-03-01",
    "end": "2023-09-01"
}
```

```python
response_visits = requests.get(url_visits, params=params)
response_reg = requests.get(url_reg, params=params)

data_visits = response_visits.json()
data_reg = response_reg.json()
#print(data_visits[:2])

df_visits = pd.DataFrame(data_visits)
print(df_visits.head())

print('\n --- \n')

df_reg = pd.DataFrame(data_reg)
print(df_reg.head())

print('\n --- \n')

df_ads = pd.read_csv('./ads.csv') 
print(df_ads.head())
```

```python
# --- 2. Обработка визитов: оставляем только последний визит для каждого visit_id ---
# Сначала приводим datetime к типу datetime
df_visits['datetime'] = pd.to_datetime(df_visits['datetime'])

# Сортируем по дате, чтобы последний был в конце
df_visits_sorted = df_visits.sort_values('datetime')

# Оставляем только последний визит на visit_id
df_visits_unique = df_visits_sorted.drop_duplicates(subset='visit_id', keep='last')


# --- 3. Агрегация визитов по дате и платформе ---
# Создаём поле date_group как дату без времени (YYYY-MM-DD)
df_visits_unique['date_group'] = df_visits_unique['datetime'].dt.date

visits_agg = (
    df_visits_unique
    .groupby(['date_group', 'platform'], observed=True)
    .size()
    .reset_index(name='visits')
)


# --- 4. Агрегация регистраций по дате и платформе ---
df_reg['datetime'] = pd.to_datetime(df_reg['datetime'])
df_reg['date_group'] = df_reg['datetime'].dt.date

reg_agg = (
    df_reg
    .groupby(['date_group', 'platform'], observed=True)
    .size()
    .reset_index(name='registrations')
)


# --- 5. Объединение и расчёт конверсии ---
# Объединяем по date_group и platform (LEFT JOIN, чтобы не потерять дни без регистраций)
df_merged = visits_agg.merge(reg_agg, on=['date_group', 'platform'], how='left')

# Заполняем пропуски нулями (если визиты были, а регистраций нет)
df_merged['registrations'] = df_merged['registrations'].fillna(0).astype(int)

# Расчёт конверсии: registrations / visits. Если визитов 0 — конверсия 0
df_merged['conversion'] = (
    df_merged.apply(lambda row: row['registrations'] / row['visits'] if row['visits'] > 0 else 0, axis=1)
)

# Приводим date_group к строке, чтобы JSON был читаемым (дата как строка)
df_merged['date_group'] = df_merged['date_group'].astype(str)


# --- 6. Сохранение в conversion.json ---
df_merged.to_json('./conversion_dict.json', orient='records', force_ascii=False)

print(df_merged.head())
```

```python
# Обработка визитов: последний визит на visit_id
df_visits['datetime'] = pd.to_datetime(df_visits['datetime'])
df_visits_unique = (
    df_visits
    .sort_values('datetime')
    .drop_duplicates(subset='visit_id', keep='last')
    .copy()
)

# ИСПРАВЛЕНИЕ: группируем по дням, обнуляя время, но сохраняя тип datetime
df_visits_unique['date_group'] = df_visits_unique['datetime'].dt.normalize()

visits_agg = (
    df_visits_unique
    .groupby(['date_group', 'platform'], observed=True)
    .size()
    .reset_index(name='visits')
)

# Агрегация регистраций по дням и платформе
df_reg['datetime'] = pd.to_datetime(df_reg['datetime'])
# ИСПРАВЛЕНИЕ: тоже нормализуем, чтобы дни совпадали
df_reg['date_group'] = df_reg['datetime'].dt.normalize()

reg_agg = (
    df_reg
    .groupby(['date_group', 'platform'], observed=True)
    .size()
    .reset_index(name='registrations')
)

# Объединение и конверсия
df_merged = visits_agg.merge(reg_agg, on=['date_group', 'platform'], how='left')
df_merged['registrations'] = df_merged['registrations'].fillna(0).astype(int)

# Векторизованный расчёт конверсии (быстрее и чище, чем apply)
df_merged['conversion'] = df_merged['registrations'] / df_merged['visits']
df_merged['conversion'] = (df_merged['conversion'] * 100).fillna(0)

# Сохранение в JSON (timestamp-даты, orient=columns)
df_merged.to_json('./conversion.json', orient='columns', force_ascii=False)

print(df_merged[['date_group', 'platform', 'visits', 'registrations', 'conversion']].head(15))
```

```python
# --- Подготовка рекламы (агрегация по дате) ---
df_ads['date'] = pd.to_datetime(df_ads['date'])
#df_ads['date_group'] = df_ads['date'].dt.date
# Делаем date_group как datetime (полночь дня) вместо date
df_ads['date_group'] = df_ads['date'].dt.normalize()  # это даст datetime64[ns], время = 00:00:00


ads_agg = (
    df_ads
    .groupby('date_group', observed=True)
    .agg(
        cost=('cost', 'sum'),
        utm_campaign=('utm_campaign', lambda x: x.iloc[0] if len(x) > 0 else 'none')
    )
    .reset_index()
)

ads_agg = ads_agg.sort_values('date_group')

# ---  Объединение с конверсиями ---
df_final = df_merged.merge(ads_agg, on='date_group', how='left')

# Заполняем отсутствующие значения (если в день не было рекламы)
df_final['cost'] = df_final['cost'].fillna(0)
df_final['utm_campaign'] = df_final['utm_campaign'].fillna('none')

# Сортировка: сначала по дате, потом по платформе
df_final = df_final.sort_values(['date_group', 'platform'])
print(df_final.head())

# --- Сохранение в JSON ---
df_final.to_json('./ads.json', orient='records', force_ascii=False)
```

## Итоговые визиты (сумма по всем платформам)

```python
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 12

# --- Фильтрация по датам ---
start_date = '2023-03-01'
end_date = '2023-04-30'

df_period = df_final[(df_final['date_group'] >= start_date) & (df_final['date_group'] <= end_date)]

# Группируем по дате и суммируем визиты
visits_by_date = df_period.groupby('date_group')['visits'].sum()

# Строим график
plt.figure(figsize=(14, 6))
bars = plt.bar(visits_by_date.index, visits_by_date.values, color='#93c9df')

# Добавляем подписи над столбцами
plt.bar_label(bars, padding=3, fontsize=9, color='black', fmt='%d')

# Убираем вертикальные сетки
ax = plt.gca()
ax.grid(axis='y', linestyle='--', alpha=0.3)  # оставляем только горизонтальные (опционально)
ax.grid(axis='x', visible=False)              # отключаем вертикальные

# Настраиваем ось X: каждый день, без пропусков, наклон 45°
ax.xaxis.set_major_locator(mdates.DayLocator(interval=1))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
plt.xticks(rotation=45, ha='right')  # ha='right' — чтобы текст не «улетал» за край

plt.title("Total Visits")
plt.xlabel("Date")
plt.ylabel("visits")
#plt.xticks(rotation=45)
plt.tight_layout()

# Сохраняем и показываем
plt.savefig("./charts/total_visits.png", dpi=300)
plt.show()
```

## Итоговые визиты с разбивкой по платформам (web, android, ios)

```python
# Настройка стилей
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (14, 7)
plt.rcParams["font.size"] = 12

# Создаём папку (раскомментируй, если нужно)
#os.makedirs("./charts", exist_ok=True)

# --- 1. Фильтрация по периоду ---
start_date = '2023-03-01'
end_date = '2023-04-30'

# ВАЖНО: приводим date_group к datetime, если ещё не сделано
if not pd.api.types.is_datetime64_any_dtype(df_final['date_group']):
    df_final['date_group'] = pd.to_datetime(df_final['date_group'])

# Теперь фильтрация по датам будет работать корректно
df_filtered = df_final[
    (df_final['date_group'] >= start_date) &
    (df_final['date_group'] <= end_date)
].copy()

# --- 2. Фильтрация платформ ---
allowed_platforms = ['web', 'android', 'ios']
df_plot = df_filtered[df_filtered['platform'].isin(allowed_platforms)].copy()

# Агрегация: суммируем визиты по дате и платформе
visits_pivot = df_plot.pivot_table(
    index='date_group',
    columns='platform',
    values='visits',
    aggfunc='sum'
)

# Сортируем индекс (даты)
visits_pivot = visits_pivot.sort_index()

# --- 3. Построение столбчатой диаграммы ---
ax = visits_pivot.plot(
    kind='bar',
    stacked=True,
    figsize=(14, 7),
    width=0.8,
    edgecolor='black',
    linewidth=0.5
)

# Настраиваем ось X: только если индекс — это datetime
# (для bar-графика mdates не обязателен, но полезен, если хочешь именно даты)
# Для bar-графика лучше просто форматировать метки, а не ставить DayLocator
ax.set_xticklabels(
    [ts.strftime('%Y-%m-%d') for ts in visits_pivot.index],
    rotation=45,
    ha="right"
)

plt.title("Итоговые визиты с разбивкой по платформам (web, android, ios)", fontsize=14)
plt.xlabel("Дата", fontsize=12)
plt.ylabel("Визиты", fontsize=12)
plt.legend(title="Платформа", loc='upper right')
plt.tight_layout()

# Сохраняем график
plt.savefig("./charts/visits_by_platform_bar.png", dpi=300)
#print("График сохранён: ./charts/visits_by_platform_bar.png")
plt.show()
```

## Итоговые конверсии

```python
# 1. Исключаем платформу 'bot'
df_no_bot = df_merged[df_merged['platform'] != 'bot'].copy()

# 2. Агрегируем по дате
df_daily = (
    df_no_bot
    .groupby('date_group', as_index=False)[['visits', 'registrations']]
    .sum()
)

# Итоговая конверсия
df_daily['conversion_total'] = (df_daily['registrations'] / df_daily['visits']) * 100
df_daily['conversion_total'] = df_daily['conversion_total'].round(2)

# Приводим date_group к datetime ДО фильтрации
df_daily['date_group'] = pd.to_datetime(df_daily['date_group'])

# 3. Фильтруем по периоду
start_date = '2023-03-01'
end_date = '2023-04-30'
mask = (df_daily['date_group'] >= start_date) & (df_daily['date_group'] <= end_date)
df_plot = df_daily.loc[mask].copy()

# 4. Строим график и добавляем подписи
fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(
    df_plot['date_group'],
    df_plot['conversion_total'],
    marker='o',
    linewidth=2,
    color='#2E86AB'
)

# Подписи над точками
for i, row in df_plot.iterrows():
    x_val = row['date_group']
    y_val = row['conversion_total']

    # Пропускаем битые значения
    if pd.isna(x_val) or pd.isna(y_val) or np.isinf(y_val):
        continue

    ax.annotate(
        f'{y_val:.1f}%',
        xy=(x_val, y_val),          # координаты точки
        xytext=(0, 8),              # сдвиг вверх на 8 пунктов
        textcoords='offset points',
        ha='center',
        va='bottom',
        fontsize=9,
        color='#2E86AB',
        weight='bold'
    )

ax.set_title('Итоговые конверсии по дням (без bot)', fontsize=14)
ax.set_xlabel('Дата', fontsize=12)
ax.set_ylabel('Конверсия, %', fontsize=12)
ax.grid(True, linestyle='--', alpha=0.3)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("./charts/total_conversion", dpi=300)
plt.show()
```

## Визиты за весь период с цветовым выделением рекламной кампании

```python
plt.figure(figsize=(12, 6))
# Сортируем, чтобы порядок был предсказуемым
df_plot = df_final.sort_values('date_group')

scatter = plt.scatter(
    df_plot['date_group'],
    df_plot['visits'],
    c=df_plot['utm_campaign'].astype('category').cat.codes,  # кодируем кампании в цвета
    cmap='tab10',
    edgecolors='k',
    linewidth=0.5,
    alpha=0.8
)

# Добавляем легенду с названиями кампаний
handles, labels = scatter.legend_elements()
legend_labels = [
    df_plot['utm_campaign'].cat.categories[i] if i < len(df_plot['utm_campaign'].cat.categories) else 'none'
    for i in [int(h.get_label()) for h in handles]
]
# Но проще сделать легенду вручную по уникальным кампаниям
unique_campaigns = df_plot['utm_campaign'].unique()
colors = plt.get_cmap('tab10')(range(len(unique_campaigns)))

from matplotlib.patches import Patch
legend_handles = [Patch(color=colors[i], label=unique_campaigns[i]) for i in range(len(unique_campaigns))]
plt.legend(handles=legend_handles, title="Кампания")

plt.title("Визиты с выделением рекламных кампаний")
plt.xlabel("Дата")
plt.ylabel("Визиты")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("./charts/visits_with_campaign.png", dpi=300)
plt.show()
```
